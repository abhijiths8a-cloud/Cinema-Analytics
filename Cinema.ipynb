```python
import sqlite3

database = sqlite3.connect("cinema.db")

cursor = database.cursor()
```

```python
# Enable foreign key constraints
cursor.execute("PRAGMA foreign_keys = ON")

# ---------------- CREATE DIRECTORS TABLE ----------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS directors (
    director_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    nationality TEXT
)
""")

# ---------------- CREATE MOVIES TABLE ----------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS movies (
    movie_id INTEGER PRIMARY KEY AUTOINCREMENT,
    director_id INTEGER,
    title TEXT NOT NULL,
    release_year INTEGER,
    FOREIGN KEY (director_id)
    REFERENCES directors(director_id)
    ON DELETE CASCADE
)
""")

# ---------------- CREATE BOX OFFICE TABLE ----------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS box_office (
    record_id INTEGER PRIMARY KEY AUTOINCREMENT,
    movie_id INTEGER,
    revenue_millions REAL,
    territory TEXT,
    FOREIGN KEY (movie_id)
    REFERENCES movies(movie_id)
    ON DELETE CASCADE
)
""")

# ---------------- INSERT SAMPLE DIRECTORS ----------------

cursor.execute("SELECT COUNT(*) FROM directors")

if cursor.fetchone()[0] == 0:

    directors_data = [
        ("Christopher Nolan", "British-American"),
        ("S. S. Rajamouli", "Indian"),
        ("Steven Spielberg", "American"),
        ("James Cameron", "Canadian")
    ]

    cursor.executemany("""
    INSERT INTO directors (name, nationality)
    VALUES (?, ?)
    """, directors_data)

    # ---------------- INSERT SAMPLE MOVIES ----------------

    movies_data = [
        (1, "Inception", 2010),
        (1, "Interstellar", 2014),
        (2, "RRR", 2022),
        (2, "Baahubali", 2015),
        (3, "Jurassic Park", 1993),
        (4, "Avatar", 2009)
    ]

    cursor.executemany("""
    INSERT INTO movies (director_id, title, release_year)
    VALUES (?, ?, ?)
    """, movies_data)

    # ---------------- INSERT BOX OFFICE RECORDS ----------------

    box_office_data = [
        (1, 839.0, "Worldwide"),
        (2, 731.0, "Worldwide"),
        (3, 160.0, "Worldwide"),
        (4, 650.0, "Worldwide"),
        (5, 1100.0, "Worldwide"),
        (6, 2923.0, "Worldwide")
    ]

    cursor.executemany("""
    INSERT INTO box_office (movie_id, revenue_millions, territory)
    VALUES (?, ?, ?)
    """, box_office_data)

# Save changes
database.commit()

print("Database and tables created successfully!")
print("Sample records inserted successfully!")
```

```python
# ---------------- INSERT DIRECTOR ----------------

def add_director(name, nationality):
    sql = """
    INSERT INTO directors (name, nationality)
    VALUES (?, ?)
    """
    params = (name, nationality)

    cursor.execute(sql, params)
    database.commit()


# ---------------- INSERT MOVIE ----------------

def add_movie(director_id, title, release_year):
    sql = """
    INSERT INTO movies (director_id, title, release_year)
    VALUES (?, ?, ?)
    """
    params = (director_id, title, release_year)

    cursor.execute(sql, params)
    database.commit()


# ---------------- INSERT BOX OFFICE RECORD ----------------

def add_box_office(movie_id, revenue, territory):
    sql = """
    INSERT INTO box_office (movie_id, revenue_millions, territory)
    VALUES (?, ?, ?)
    """
    params = (movie_id, revenue, territory)

    cursor.execute(sql, params)
    database.commit()


# ---------------- DISPLAY DIRECTORS ----------------

def get_directors():
    sql = "SELECT * FROM directors"
    cursor.execute(sql)
    return cursor.fetchall()


# ---------------- DISPLAY MOVIES ----------------

def get_movies():
    sql = "SELECT * FROM movies"
    cursor.execute(sql)
    return cursor.fetchall()


# ---------------- DISPLAY BOX OFFICE ----------------

def get_box_office():
    sql = "SELECT * FROM box_office"
    cursor.execute(sql)
    return cursor.fetchall()


# ---------------- DELETE DIRECTOR ----------------

def delete_director(director_id):
    sql = "DELETE FROM directors WHERE director_id = ?"
    cursor.execute(sql, (director_id,))
    database.commit()


# ---------------- DELETE MOVIE ----------------

def delete_movie(movie_id):
    sql = "DELETE FROM movies WHERE movie_id = ?"
    cursor.execute(sql, (movie_id,))
    database.commit()


print("All database functions created successfully!")
```

```python
def get_filmography():
    sql = """
    SELECT
        directors.name,
        movies.title,
        movies.release_year,
        box_office.revenue_millions,
        box_office.territory
    FROM directors
    INNER JOIN movies
        ON directors.director_id = movies.director_id
    LEFT JOIN box_office
        ON movies.movie_id = box_office.movie_id
    ORDER BY directors.name;
    """

    cursor.execute(sql)
    return cursor.fetchall()


print("COMPLETE FILMOGRAPHY:")

for record in get_filmography():
    print(record)
```

```python
# Update Director
def update_director(director_id, name, nationality):
    sql = """
    UPDATE directors
    SET name = ?, nationality = ?
    WHERE director_id = ?
    """
    cursor.execute(sql, (name, nationality, director_id))
    database.commit()


# Update Movie
def update_movie(movie_id, director_id, title, release_year):
    sql = """
    UPDATE movies
    SET director_id = ?, title = ?, release_year = ?
    WHERE movie_id = ?
    """
    cursor.execute(sql, (director_id, title, release_year, movie_id))
    database.commit()


# Update Box Office Record
def update_box_office(record_id, movie_id, revenue, territory):
    sql = """
    UPDATE box_office
    SET movie_id = ?, revenue_millions = ?, territory = ?
    WHERE record_id = ?
    """
    cursor.execute(sql, (movie_id, revenue, territory, record_id))
    database.commit()


print("All update functions created successfully!")
```

```python

# ---------------- EDIT RECORDS WINDOW ----------------

def open_edit_window():

    edit_window = tk.Toplevel(root)
    edit_window.title("Edit Database Records")
    edit_window.geometry("550x550")
    edit_window.configure(bg=BG)

    tk.Label(
        edit_window,
        text="EDIT DATABASE RECORDS",
        font=("Arial", 17, "bold"),
        bg=NAVY,
        fg=WHITE,
        pady=15
    ).pack(fill="x")

    form = tk.Frame(edit_window, bg=WHITE, padx=25, pady=20)
    form.pack(fill="both", expand=True, padx=25, pady=20)

    tk.Label(form, text="Select Record Type", bg=WHITE).pack(anchor="w")

    record_type = ttk.Combobox(
        form,
        values=["Director", "Movie", "Box Office"],
        state="readonly"
    )
    record_type.pack(fill="x", pady=5)
    record_type.set("Director")

    tk.Label(form, text="Record ID", bg=WHITE).pack(anchor="w", pady=(10, 0))
    record_id = tk.Entry(form)
    record_id.pack(fill="x", pady=5)

    fields_frame = tk.Frame(form, bg=WHITE)
    fields_frame.pack(fill="both", expand=True, pady=10)

    entries = {}

    def load_record():

        for widget in fields_frame.winfo_children():
            widget.destroy()

        entries.clear()

        try:
            rid = int(record_id.get())
            kind = record_type.get()

            if kind == "Director":
                cursor.execute(
                    "SELECT name, nationality FROM directors WHERE director_id=?",
                    (rid,)
                )
                record = cursor.fetchone()
                labels = ["Name", "Nationality"]

            elif kind == "Movie":
                cursor.execute(
                    "SELECT director_id, title, release_year FROM movies WHERE movie_id=?",
                    (rid,)
                )
                record = cursor.fetchone()
                labels = ["Director ID", "Movie Title", "Release Year"]

            else:
                cursor.execute(
                    """SELECT movie_id, revenue_millions, territory
                       FROM box_office WHERE record_id=?""",
                    (rid,)
                )
                record = cursor.fetchone()
                labels = ["Movie ID", "Revenue (millions)", "Territory"]

            if record is None:
                messagebox.showerror("Error", "Record not found.")
                return

            for label, value in zip(labels, record):
                tk.Label(
                    fields_frame,
                    text=label,
                    bg=WHITE
                ).pack(anchor="w", pady=(8, 2))

                entry = tk.Entry(fields_frame)
                entry.insert(0, str(value))
                entry.pack(fill="x", pady=3)

                entries[label] = entry

        except ValueError:
            messagebox.showerror("Error", "Enter a valid numeric record ID.")

    def save_changes():

        try:
            rid = int(record_id.get())
            kind = record_type.get()
            values = {key: entry.get().strip()
                      for key, entry in entries.items()}

            if not all(values.values()):
                messagebox.showerror("Error", "All fields are required.")
                return

            if kind == "Director":
                update_director(
                    rid,
                    values["Name"],
                    values["Nationality"]
                )

            elif kind == "Movie":
                director_id = int(values["Director ID"])
                year = int(values["Release Year"])

                if year <= 0:
                    raise ValueError

                update_movie(
                    rid,
                    director_id,
                    values["Movie Title"],
                    year
                )

            else:
                movie_id = int(values["Movie ID"])
                revenue = float(values["Revenue (millions)"])

                if revenue < 0:
                    raise ValueError

                update_box_office(
                    rid,
                    movie_id,
                    revenue,
                    values["Territory"]
                )

            messagebox.showinfo("Success", "Record updated successfully!")

            refresh_dropdowns()
            refresh_table()

            edit_window.destroy()

        except ValueError:
            messagebox.showerror(
                "Validation Error",
                "Check the numeric values and enter valid data."
            )

        except Exception as error:
            messagebox.showerror("Database Error", str(error))

    action_button(form, "LOAD RECORD", load_record).pack(pady=10)
    action_button(form, "SAVE CHANGES", save_changes).pack(pady=10)


# ---------------- ADD EDIT BUTTON ----------------

action_button(
    controls,
    "EDIT RECORDS",
    open_edit_window
).pack(side="left", padx=10)

print("Edit Records functionality added successfully!")
```

```python

# ============================================================
# CINEMA ANALYTICS - FINAL GUI
# Add, View, Edit and Delete Directors, Movies and Box Office
# ============================================================

import tkinter as tk
from tkinter import ttk, messagebox
import sqlite3

# Reuse the existing database connection
cursor.execute("PRAGMA foreign_keys = ON")

# -------------------- WINDOW SETUP --------------------

root = tk.Tk()
root.title("Cinema Analytics | Filmography & Box Office Tracker")
root.geometry("1100x700")
root.configure(bg="#F3F6FC")

NAVY = "#14213D"
BLUE = "#2563EB"
LIGHT = "#F3F6FC"
WHITE = "#FFFFFF"
TEXT = "#1E293B"
MUTED = "#64748B"

style = ttk.Style()
style.theme_use("clam")

style.configure(
    "TNotebook",
    background=LIGHT,
    borderwidth=0
)

style.configure(
    "TNotebook.Tab",
    padding=(22, 12),
    font=("Arial", 10, "bold")
)

style.configure(
    "Treeview",
    background=WHITE,
    foreground=TEXT,
    rowheight=30,
    fieldbackground=WHITE,
    font=("Arial", 10)
)

style.configure(
    "Treeview.Heading",
    background=NAVY,
    foreground=WHITE,
    font=("Arial", 10, "bold")
)

style.map(
    "Treeview",
    background=[("selected", "#DBEAFE")],
    foreground=[("selected", TEXT)]
)

# -------------------- HELPER FUNCTIONS --------------------

def make_label(parent, text):
    return tk.Label(
        parent,
        text=text,
        bg=WHITE,
        fg=TEXT,
        font=("Arial", 10, "bold")
    )

def make_entry(parent):
    return tk.Entry(
        parent,
        font=("Arial", 11),
        relief="solid",
        bd=1
    )

def make_button(parent, text, command, color=BLUE):
    return tk.Button(
        parent,
        text=text,
        command=command,
        bg=color,
        fg=WHITE,
        activebackground=NAVY,
        activeforeground=WHITE,
        font=("Arial", 10, "bold"),
        relief="flat",
        padx=15,
        pady=8,
        cursor="hand2"
    )

def get_director_options():
    cursor.execute(
        "SELECT director_id, name FROM directors ORDER BY name"
    )
    return [
        f"{row[0]} - {row[1]}"
        for row in cursor.fetchall()
    ]

def get_movie_options():
    cursor.execute(
        "SELECT movie_id, title FROM movies ORDER BY title"
    )
    return [
        f"{row[0]} - {row[1]}"
        for row in cursor.fetchall()
    ]

def extract_id(value):
    return int(value.split(" - ")[0])

# -------------------- HEADER --------------------

header = tk.Frame(root, bg=NAVY, height=105)
header.pack(fill="x")

tk.Label(
    header,
    text="CINEMA ANALYTICS",
    bg=NAVY,
    fg=WHITE,
    font=("Arial", 23, "bold")
).pack(anchor="w", padx=30, pady=(20, 2))

tk.Label(
    header,
    text="DIRECTOR FILMOGRAPHY & BOX OFFICE TRACKER",
    bg=NAVY,
    fg="#BFDBFE",
    font=("Arial", 10)
).pack(anchor="w", padx=32)

# -------------------- NOTEBOOK --------------------

notebook = ttk.Notebook(root)
notebook.pack(fill="both", expand=True, padx=20, pady=20)

director_tab = tk.Frame(notebook, bg=WHITE)
movie_tab = tk.Frame(notebook, bg=WHITE)
box_tab = tk.Frame(notebook, bg=WHITE)
records_tab = tk.Frame(notebook, bg=WHITE)

notebook.add(director_tab, text="DIRECTORS")
notebook.add(movie_tab, text="MOVIES")
notebook.add(box_tab, text="BOX OFFICE")
notebook.add(records_tab, text="DATABASE RECORDS")

# -------------------- DIRECTOR TAB --------------------

tk.Label(
    director_tab,
    text="ADD NEW DIRECTOR",
    bg=WHITE,
    fg=NAVY,
    font=("Arial", 17, "bold")
).pack(anchor="w", padx=35, pady=(35, 20))

director_form = tk.Frame(director_tab, bg=WHITE)
director_form.pack(anchor="w", padx=35)

make_label(director_form, "Director Name").grid(
    row=0, column=0, sticky="w", pady=8
)
director_name = make_entry(director_form)
director_name.grid(row=0, column=1, padx=20, pady=8)

make_label(director_form, "Nationality").grid(
    row=1, column=0, sticky="w", pady=8
)
director_nationality = make_entry(director_form)
director_nationality.grid(row=1, column=1, padx=20, pady=8)

def add_director_gui():
    name = director_name.get().strip()
    nationality = director_nationality.get().strip()

    if not name:
        messagebox.showwarning("Input Required", "Enter the director's name.")
        return

    try:
        add_director(name, nationality)
        messagebox.showinfo("Success", "Director added successfully.")

        director_name.delete(0, tk.END)
        director_nationality.delete(0, tk.END)

        refresh_dropdowns()
        refresh_table()

    except Exception as error:
        messagebox.showerror("Error", str(error))

make_button(
    director_tab,
    "ADD DIRECTOR",
    add_director_gui
).pack(anchor="w", padx=35, pady=25)

# -------------------- MOVIE TAB --------------------

tk.Label(
    movie_tab,
    text="ADD NEW MOVIE",
    bg=WHITE,
    fg=NAVY,
    font=("Arial", 17, "bold")
).pack(anchor="w", padx=35, pady=(30, 20))

movie_form = tk.Frame(movie_tab, bg=WHITE)
movie_form.pack(anchor="w", padx=35)

make_label(movie_form, "Movie Title").grid(
    row=0, column=0, sticky="w", pady=8
)
movie_title = make_entry(movie_form)
movie_title.grid(row=0, column=1, padx=20, pady=8)

make_label(movie_form, "Release Year").grid(
    row=1, column=0, sticky="w", pady=8
)
movie_year = make_entry(movie_form)
movie_year.grid(row=1, column=1, padx=20, pady=8)

make_label(movie_form, "Director").grid(
    row=2, column=0, sticky="w", pady=8
)
movie_director = ttk.Combobox(
    movie_form,
    width=27,
    state="readonly"
)
movie_director.grid(row=2, column=1, padx=20, pady=8)

def add_movie_gui():
    title = movie_title.get().strip()
    year = movie_year.get().strip()
    director = movie_director.get()

    if not title or not year or not director:
        messagebox.showwarning("Input Required", "Complete all movie fields.")
        return

    try:
        year = int(year)
        director_id = extract_id(director)

        add_movie(director_id, title, year)

        messagebox.showinfo("Success", "Movie added successfully.")

        movie_title.delete(0, tk.END)
        movie_year.delete(0, tk.END)

        refresh_dropdowns()
        refresh_table()

    except ValueError:
        messagebox.showerror("Invalid Input", "Enter a valid release year.")
    except Exception as error:
        messagebox.showerror("Error", str(error))

make_button(
    movie_tab,
    "ADD MOVIE",
    add_movie_gui
).pack(anchor="w", padx=35, pady=25)

# -------------------- BOX OFFICE TAB --------------------

tk.Label(
    box_tab,
    text="ADD BOX OFFICE RECORD",
    bg=WHITE,
    fg=NAVY,
    font=("Arial", 17, "bold")
).pack(anchor="w", padx=35, pady=(30, 20))

box_form = tk.Frame(box_tab, bg=WHITE)
box_form.pack(anchor="w", padx=35)

make_label(box_form, "Movie").grid(
    row=0, column=0, sticky="w", pady=8
)
box_movie = ttk.Combobox(
    box_form,
    width=27,
    state="readonly"
)
box_movie.grid(row=0, column=1, padx=20, pady=8)

make_label(box_form, "Revenue (Millions)").grid(
    row=1, column=0, sticky="w", pady=8
)
box_revenue = make_entry(box_form)
box_revenue.grid(row=1, column=1, padx=20, pady=8)

make_label(box_form, "Territory").grid(
    row=2, column=0, sticky="w", pady=8
)
box_territory = make_entry(box_form)
box_territory.grid(row=2, column=1, padx=20, pady=8)

def add_box_gui():
    movie = box_movie.get()
    revenue = box_revenue.get().strip()
    territory = box_territory.get().strip()

    if not movie or not revenue or not territory:
        messagebox.showwarning("Input Required", "Complete all box office fields.")
        return

    try:
        movie_id = extract_id(movie)
        revenue = float(revenue)

        if revenue < 0:
            raise ValueError

        add_box_office(movie_id, revenue, territory)

        messagebox.showinfo("Success", "Box office record added successfully.")

        box_revenue.delete(0, tk.END)
        box_territory.delete(0, tk.END)

        refresh_table()

    except ValueError:
        messagebox.showerror(
            "Invalid Input",
            "Enter a valid non-negative revenue value."
        )
    except Exception as error:
        messagebox.showerror("Error", str(error))

make_button(
    box_tab,
    "ADD BOX OFFICE RECORD",
    add_box_gui
).pack(anchor="w", padx=35, pady=25)

# -------------------- DATABASE RECORDS TAB --------------------

tk.Label(
    records_tab,
    text="COMPLETE FILMOGRAPHY",
    bg=WHITE,
    fg=NAVY,
    font=("Arial", 17, "bold")
).pack(anchor="w", padx=20, pady=(20, 10))

table_frame = tk.Frame(records_tab, bg=WHITE)
table_frame.pack(fill="both", expand=True, padx=20, pady=10)

columns = (
    "Director",
    "Movie",
    "Year",
    "Revenue",
    "Territory"
)

tree = ttk.Treeview(
    table_frame,
    columns=columns,
    show="headings",
    height=12
)

for column in columns:
    tree.heading(column, text=column)
    tree.column(column, width=150, anchor="center")

tree.pack(side="left", fill="both", expand=True)

scrollbar = ttk.Scrollbar(
    table_frame,
    orient="vertical",
    command=tree.yview
)
scrollbar.pack(side="right", fill="y")

tree.configure(yscrollcommand=scrollbar.set)

def refresh_table():
    for item in tree.get_children():
        tree.delete(item)

    for record in get_filmography():
        tree.insert("", tk.END, values=record)

def refresh_dropdowns():
    movie_director["values"] = get_director_options()
    box_movie["values"] = get_movie_options()

    if movie_director["values"]:
        movie_director.current(0)

    if box_movie["values"]:
        box_movie.current(0)

# -------------------- EDIT RECORDS WINDOW --------------------

def open_edit_window():

    edit_window = tk.Toplevel(root)
    edit_window.title("Edit Database Records")
    edit_window.geometry("520x540")
    edit_window.configure(bg=WHITE)

    tk.Label(
        edit_window,
        text="EDIT DATABASE RECORDS",
        bg=WHITE,
        fg=NAVY,
        font=("Arial", 16, "bold")
    ).pack(pady=20)

    form = tk.Frame(edit_window, bg=WHITE)
    form.pack(padx=25, pady=10, fill="x")

    make_label(form, "Record Type").grid(
        row=0, column=0, sticky="w", pady=8
    )

    record_type = ttk.Combobox(
        form,
        values=["Director", "Movie", "Box Office"],
        state="readonly",
        width=25
    )
    record_type.grid(row=0, column=1, padx=15, pady=8)
    record_type.current(0)

    make_label(form, "Record ID").grid(
        row=1, column=0, sticky="w", pady=8
    )

    record_id = make_entry(form)
    record_id.grid(row=1, column=1, padx=15, pady=8)

    fields_frame = tk.Frame(edit_window, bg=WHITE)
    fields_frame.pack(padx=25, pady=10, fill="x")

    entries = {}

    def build_fields():
        for widget in fields_frame.winfo_children():
            widget.destroy()

        entries.clear()

        selected = record_type.get()

        if selected == "Director":
            fields = ["Name", "Nationality"]
        elif selected == "Movie":
            fields = ["Director ID", "Title", "Release Year"]
        else:
            fields = ["Movie ID", "Revenue (Millions)", "Territory"]

        for row, field in enumerate(fields):
            make_label(fields_frame, field).grid(
                row=row, column=0, sticky="w", pady=8
            )

            entry = make_entry(fields_frame)
            entry.grid(row=row, column=1, padx=15, pady=8)
            entries[field] = entry

    def load_record():
        try:
            rid = int(record_id.get())
            selected = record_type.get()

            if selected == "Director":
                cursor.execute(
                    "SELECT name, nationality FROM directors WHERE director_id=?",
                    (rid,)
                )
            elif selected == "Movie":
                cursor.execute(
                    "SELECT director_id, title, release_year FROM movies WHERE movie_id=?",
                    (rid,)
                )
            else:
                cursor.execute(
                    """SELECT movie_id, revenue_millions, territory
                       FROM box_office WHERE record_id=?""",
                    (rid,)
                )

            data = cursor.fetchone()

            if data is None:
                messagebox.showwarning(
                    "Not Found",
                    "No record exists with this ID.",
                    parent=edit_window
                )
                return

            for entry, value in zip(entries.values(), data):
                entry.delete(0, tk.END)
                entry.insert(0, str(value))

        except ValueError:
            messagebox.showerror(
                "Invalid ID",
                "Enter a numeric record ID.",
                parent=edit_window
            )

    def save_record():
        try:
            rid = int(record_id.get())
            selected = record_type.get()
            values = [entry.get().strip() for entry in entries.values()]

            if any(value == "" for value in values):
                messagebox.showwarning(
                    "Input Required",
                    "Complete all fields.",
                    parent=edit_window
                )
                return

            if selected == "Director":
                update_director(rid, values[0], values[1])

            elif selected == "Movie":
                update_movie(
                    rid,
                    int(values[0]),
                    values[1],
                    int(values[2])
                )

            else:
                revenue = float(values[1])

                if revenue < 0:
                    raise ValueError

                update_box_office(
                    rid,
                    int(values[0]),
                    revenue,
                    values[2]
                )

            messagebox.showinfo(
                "Success",
                "Record updated successfully.",
                parent=edit_window
            )

            refresh_dropdowns()
            refresh_table()
            edit_window.destroy()

        except ValueError:
            messagebox.showerror(
                "Invalid Input",
                "Check the IDs, year and revenue values.",
                parent=edit_window
            )
        except sqlite3.IntegrityError:
            messagebox.showerror(
                "Database Error",
                "The referenced Director ID or Movie ID does not exist.",
                parent=edit_window
            )
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=edit_window)

    record_type.bind("<<ComboboxSelected>>", lambda event: build_fields())

    make_button(
        form,
        "LOAD RECORD",
        load_record
    ).grid(row=2, column=1, pady=15, sticky="w")

    build_fields()

    make_button(
        edit_window,
        "SAVE CHANGES",
        save_record
    ).pack(pady=20)

# -------------------- DELETE RECORDS --------------------

def delete_selected_record():
    selected = tree.selection()

    if not selected:
        messagebox.showwarning(
            "Select Record",
            "Select a filmography row first."
        )
        return

    values = tree.item(selected[0], "values")
    movie_title_value = values[1]

    cursor.execute(
        "SELECT movie_id FROM movies WHERE title=?",
        (movie_title_value,)
    )
    result = cursor.fetchone()

    if result is None:
        messagebox.showerror("Error", "Movie record not found.")
        return

    movie_id = result[0]

    confirm = messagebox.askyesno(
        "Confirm Delete",
        f"Delete '{movie_title_value}' and its box office records?"
    )

    if confirm:
        try:
            delete_movie(movie_id)
            messagebox.showinfo("Success", "Movie deleted successfully.")
            refresh_table()
            refresh_dropdowns()
        except Exception as error:
            messagebox.showerror("Error", str(error))

# -------------------- ACTION BUTTONS --------------------

action_frame = tk.Frame(records_tab, bg=WHITE)
action_frame.pack(pady=15)

make_button(
    action_frame,
    "EDIT RECORDS",
    open_edit_window
).pack(side="left", padx=10)

make_button(
    action_frame,
    "DELETE SELECTED MOVIE",
    delete_selected_record,
    color="#DC2626"
).pack(side="left", padx=10)

make_button(
    action_frame,
    "REFRESH TABLE",
    refresh_table,
    color="#475569"
).pack(side="left", padx=10)

# -------------------- INITIAL REFRESH --------------------

refresh_dropdowns()
refresh_table()

# -------------------- RUN APPLICATION --------------------

root.mainloop()
```
